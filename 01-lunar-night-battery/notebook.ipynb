Personal project using only public data. Not affiliated with or endorsed by my employer.

This notebook calls the same functions as `run_all.py`. The derivation is in `EXPLAINED.md`. Set `SPACE_POWER_SIMS_MODE=smoke` to write a short run into a temporary directory and leave `figures/` alone.

```python
import os
import sys
import tempfile
from pathlib import Path

os.environ.setdefault('PYBAMM_DISABLE_TELEMETRY', '1')

here = Path.cwd().resolve()
root = here
for candidate in [here, *here.parents]:
    if (candidate / 'pyproject.toml').is_file():
        root = candidate
        break
folder = root / '01-lunar-night-battery'
for path in (root, folder):
    if str(path) not in sys.path:
        sys.path.insert(0, str(path))

from common.params import load_params
from make_figures import generate

smoke = os.environ.get('SPACE_POWER_SIMS_MODE', '').strip().lower() == 'smoke'
out = Path(tempfile.mkdtemp(prefix='lunar-night-notebook-')) if smoke else folder / 'figures'
book = load_params(folder / 'params.yaml')
summary = generate(book, out, smoke=smoke)
print(f'Wrote figures to {out}')
print('smoke' if smoke else f"relative drop 0 C vs 25 C: {summary['B']['relative_drop']:.3f}")
```
